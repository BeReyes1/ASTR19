```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 1, 100)
```

```python
def sin(x):
  return np.sin(x)
```

```python
def cos(x):
  return np.cos(x)
```

```python
f, axarr = plt.subplots(1, 2)
axarr[0].plot(sin(x), x)
axarr[0].set_title('sin(x)')
axarr[1].plot(cos(x), x)
axarr[1].set_title('cos(x)')
plt.savefig('2.pdf',format="pdf", bbox_inches="tight")
```
