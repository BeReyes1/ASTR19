```python
import numpy as np
from astropy.table import Table #import the Table class
from astropy.io import ascii # ascii plain text io
from astropy.io import fits # FITS format io

#retruns the sin value of the parameter
def sin(x):
    return np.sin(x)
```

```python
#retruns the sin value of the parameter
def cos(x):
    return np.cos(x)
```

```python
#access a value from the sin(x) array
def sinvx(index):
    x = np.linspace(0,2*(np.pi),1000) 
    sinx =[]
    for j in x:
        sinx.append(sin(j))
    return sinx[index]

#access a value from the cos(x) array
def cosvx(i):
    x = np.linspace(0,2*(np.pi),1000) 
    cosx = []
    for j in x:
        cosx.append(cos(j))
    return cosx[i]

#accesses a value from the x array
def accessx(index):
    x = np.linspace(0,2*(np.pi),1000)
    return x[index]
```

```python
x = []
sinx = []
cosx = []
for i in range(10):
    x.append(accessx(i))
    sinx.append(sinvx(i))
    cosx.append(cosvx(i))
    
data = Table([x, sinx, cosx],names=['x','sin(x)','cos(x)']) #create a talbe
ascii.write(data, 'table.txt', format='commented_header', overwrite=True) # write the table to file
data_in = ascii.read('table.txt') #read the data in
print(data_in)
```
