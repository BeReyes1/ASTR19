```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#random values at 0, 1, 2, 3, 4 due to first parameter, third is probability
data = np.random.default_rng().choice(5, 1000, p=[0.5, 0.1, 0.1, 0.2, 0.1])
```

```python
width = 1
histmin = np.floor(np.min(data))
histmax = np.ceil(np.max(data))+width

bins = 100
plt.hist(data, bins=bins, alpha=0.5, edgecolor='black')
plt.xlabel('Number')
plt.ylabel('Occurences')
plt.title('Histogram')
plt.savefig('4.pdf',format="pdf", bbox_inches="tight")
```
