```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.random.uniform(0,1,1000)
```

```python
width = 1
histmin = np.floor(np.min(data))
histmax = np.ceil(np.max(data))+width

bins = 100
plt.hist(data, bins=bins, alpha=0.5, edgecolor='black')
plt.xlabel('Number')
plt.ylabel('Occurences')
plt.title('Histogram')
plt.savefig('3.pdf',format="pdf", bbox_inches="tight")
```
